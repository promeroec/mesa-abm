```python
from onebankdd import OneBankDDModel

model = OneBankDDModel(
    width=21,
    height=21,
    impatient_probability=0.5,
    rates_mode="fixed",
    consumption_mode="variable",
    initial_deposit=1.0,
    #seed=123,
)

model.run_model()

results = model.datacollector.get_model_vars_dataframe()
results.tail()
```

```python
import matplotlib.pyplot as plt

results[["totalw", "deposits", "type1w", "type2w"]].plot()
plt.xlabel("ticks (depositors served)")
plt.ylabel("Accounts")
plt.show()
```

```python
from onebankdd.model1 import OneBankDDModel1

model1 = OneBankDDModel1(
    width=21,
    height=21,
    impatient_probability=0.5,   # p(impatient)
    rates_mode="fixed",          # "fixed" or "random"
    consumption_mode="variable", # "constant" or "variable"
    initial_deposit=1.0,
    #seed=123,
)
```

```python
model1.run_model()

results1 = model1.datacollector.get_model_vars_dataframe()
results1.tail()
```

```python
import matplotlib.pyplot as plt

results1[["totalw", "deposits", "type1w", "type2w"]].plot()
plt.xlabel("ticks (depositors served)")
plt.ylabel("Accounts")
plt.title("Model 1 – Accounts")
plt.show()
```
